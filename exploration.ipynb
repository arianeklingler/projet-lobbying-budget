```python
import pandas as pd
pd.set_option("display.max_colwidth", 200)

obj = pd.read_csv("Vues_Separees/8_objets_activites.csv", sep=";")
dec = pd.read_csv("Vues_Separees/12_decisions_concernees.csv", sep=";")
res = pd.read_csv("Vues_Separees/13_ministeres_aai_api.csv", sep=";")

lois = set(dec.loc[dec["decision_concernee"].str.startswith("Lois"), "action_representation_interet_id"])
parl = set(res.loc[res["responsable_public"].str.startswith("Député"), "action_representation_interet_id"])

# on garde les activités qui visent une loi ET des parlementaires
sub = obj[obj["activite_id"].isin(lois & parl)].copy()
sub["annee"] = pd.to_datetime(sub["date_publication_activite"], errors="coerce").dt.year

print("Nombre d'activités par année de publication :")
print(sub["annee"].value_counts().sort_index())

# 15 phrases au hasard qui citent un projet ou une proposition de loi
masque = sub["objet_activite"].fillna("").str.contains("projet de loi|proposition de loi", case=False)
print()
print("Exemples de phrases :")
print(sub.loc[masque, "objet_activite"].sample(15, random_state=1).to_string())
```

```python
import os, json
from collections import Counter

# 1. Combien d'amendements par dossier ? (les plus gros d'abord)
compte = Counter()
for dossier in os.listdir("json"):
    chemin = os.path.join("json", dossier)
    if os.path.isdir(chemin):
        for d, sd, noms in os.walk(chemin):
            compte[dossier] += len(noms)
print("Nombre de dossiers :", len(compte))
for d, n in compte.most_common(12):
    print(d, n)

# 2. Structure d'un amendement du plus gros dossier (textes coupés, listes raccourcies)
def montre(x, indent=0):
    marge = "  " * indent
    if isinstance(x, dict):
        for k, v in x.items():
            if k.startswith("@"):
                continue
            if isinstance(v, (dict, list)):
                print(marge + str(k) + ":")
                montre(v, indent + 1)
            else:
                print(marge + str(k) + ": " + str(v)[:100])
    elif isinstance(x, list):
        print(marge + "[liste de " + str(len(x)) + " éléments, on montre le premier]")
        if x:
            montre(x[0], indent + 1)
    else:
        print(marge + str(x)[:100])

plus_gros = compte.most_common(1)[0][0]
for d, sd, noms in os.walk(os.path.join("json", plus_gros)):
    jsons = [n for n in noms if n.endswith(".json")]
    if jsons:
        f = os.path.join(d, jsons[0])
        break
print()
print("Exemple :", f)
with open(f, encoding="utf-8") as fh:
    montre(json.load(fh))
```

```python
import os, json, re, html
import pandas as pd

def get(d, *cles):
    # va chercher une valeur au fond du fichier, sans planter si elle manque
    for c in cles:
        if not isinstance(d, dict):
            return None
        d = d.get(c)
    return d

def nettoie(s):
    # enlève les balises HTML et les codes bizarres
    if not isinstance(s, str):
        return ""
    s = re.sub(r"<[^>]+>", " ", s)
    s = html.unescape(s)
    return re.sub(r"\s+", " ", s).strip()

lignes = []
for dossier, sd, noms in os.walk("json/DLR5L17N52428"):
    for n in noms:
        if not n.endswith(".json"):
            continue
        with open(os.path.join(dossier, n), encoding="utf-8") as fh:
            a = json.load(fh)["amendement"]
        lignes.append({
            "uid": a.get("uid"),
            "auteur_code": get(a, "signataires", "auteur", "acteurRef"),
            "type_auteur": get(a, "signataires", "auteur", "typeAuteur"),
            "groupe_code": get(a, "signataires", "auteur", "groupePolitiqueRef"),
            "article": get(a, "pointeurFragmentTexte", "division", "titre"),
            "date_depot": get(a, "cycleDeVie", "dateDepot"),
            "sort": get(a, "cycleDeVie", "sort"),
            "expose": nettoie(get(a, "corps", "contenuAuteur", "exposeSommaire")),
        })

df_am = pd.DataFrame(lignes)
df_am.to_csv("amendements_plf2026.csv", index=False)

print(df_am.shape)
print()
print(df_am["sort"].value_counts(dropna=False))
print()
print(df_am["type_auteur"].value_counts(dropna=False))
print()
print("Groupes (codes) :")
print(df_am["groupe_code"].value_counts(dropna=False))
print()
print(df_am[["groupe_code", "article", "expose"]].sample(3, random_state=1).to_string())
```
