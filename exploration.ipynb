```python
import pandas as pd

obj = pd.read_csv("Vues_Separees/8_objets_activites.csv", sep=";")
act = pd.read_csv("Vues_Separees/10_actions_menees.csv", sep=";")
dec = pd.read_csv("Vues_Separees/12_decisions_concernees.csv", sep=";")
res = pd.read_csv("Vues_Separees/13_ministeres_aai_api.csv", sep=";")

# 1. Les numéros des tables 8 et 10 se correspondent-ils ?
print("activités :", obj["activite_id"].nunique(), "numéros différents")
print("actions :", act["action_representation_interet_id"].nunique(), "numéros différents")
print("numéros en commun :", len(set(obj["activite_id"]) & set(act["action_representation_interet_id"])))

# 2. Combien d'actions visent une loi, des parlementaires, ou les deux ?
lois = set(dec.loc[dec["decision_concernee"].str.startswith("Lois"), "action_representation_interet_id"])
parl = set(res.loc[res["responsable_public"].str.startswith("Député"), "action_representation_interet_id"])
print("actions visant des lois :", len(lois))
print("actions visant des parlementaires :", len(parl))
print("les deux à la fois :", len(lois & parl))

# 3. Combien de phrases citent un texte de loi connu ?
texte = obj["objet_activite"].fillna("")
for mot in ["PLFSS", "PLF ", "projet de loi", "proposition de loi", "Egalim"]:
    print(mot, ":", texte.str.contains(mot, case=False).sum())
```

```python
import pandas as pd
pd.set_option("display.max_colwidth", 200)

obj = pd.read_csv("Vues_Separees/8_objets_activites.csv", sep=";")
dec = pd.read_csv("Vues_Separees/12_decisions_concernees.csv", sep=";")
res = pd.read_csv("Vues_Separees/13_ministeres_aai_api.csv", sep=";")

lois = set(dec.loc[dec["decision_concernee"].str.startswith("Lois"), "action_representation_interet_id"])
parl = set(res.loc[res["responsable_public"].str.startswith("Député"), "action_representation_interet_id"])

# on garde les activités qui visent une loi ET des parlementaires
sub = obj[obj["activite_id"].isin(lois & parl)].copy()
sub["annee"] = pd.to_datetime(sub["date_publication_activite"], errors="coerce").dt.year

print("Nombre d'activités par année de publication :")
print(sub["annee"].value_counts().sort_index())

# 15 phrases au hasard qui citent un projet ou une proposition de loi
masque = sub["objet_activite"].fillna("").str.contains("projet de loi|proposition de loi", case=False)
print()
print("Exemples de phrases :")
print(sub.loc[masque, "objet_activite"].sample(15, random_state=1).to_string())
```

```python
txt = sub["objet_activite"].fillna("").str.lower()

familles = {
    "PLF": r"\bplf\b|loi de finances",
    "PLFSS": r"plfss|financement de la s[ée]curit[ée] sociale",
    "Egalim": r"egalim",
}

for nom, motif in familles.items():
    sub[nom] = txt.str.contains(motif)

print(sub.groupby("annee")[list(familles)].sum())
print()
print("activités citant au moins un de ces textes :", sub[list(familles)].any(axis=1).sum(), "sur", len(sub))
```

```python
# 1. Quelle année de loi est citée dans les phrases PLF / PLFSS ?
pl = sub[sub["PLF"] | sub["PLFSS"]].copy()
pl["annee_citee"] = pl["objet_activite"].str.extract(r"(20[12]\d)")[0]
print(pl["annee_citee"].value_counts(dropna=False).sort_index())

# 2. Que contient la table des exercices ? (juste les noms de colonnes)
ex = pd.read_csv("Vues_Separees/15_exercices.csv", sep=";")
print()
print(ex.columns.tolist())
```

```python
# 1. L'année citée dans la phrase correspond-elle à l'année de l'exercice ?
ex = pd.read_csv("Vues_Separees/15_exercices.csv", sep=";")
pl2 = pl.merge(
    ex[["exercices_id", "representants_id", "annee_debut", "montant_depense_sup"]],
    on="exercices_id", how="left"
)
pl2["annee_citee"] = pd.to_numeric(pl2["annee_citee"])
pl2["ecart"] = pl2["annee_citee"] - pl2["annee_debut"]
print("Écart entre année citée et année de l'exercice :")
print(pl2["ecart"].value_counts().sort_index())

# 2. Quelles colonnes pour décrire les lobbies et leurs secteurs ?
info = pd.read_csv("Vues_Separees/1_informations_generales.csv", sep=";", low_memory=False)
print()
print("Table 1 :", info.columns.tolist())
sect = pd.read_csv("Vues_Separees/9_secteurs_activites.csv", sep=";")
print()
print("Table 9 :", sect.columns.tolist())
print(sect.head(5).to_string())
```

```python
# 1. On garde les activités dont l'année citée est cohérente (écart 0 ou 1)
b = pl2[pl2["ecart"].isin([0, 1])].copy()
b["annee_budget"] = b["annee_citee"].astype(int)

# 2. Quel texte : PLF, PLFSS ou les deux ?
b["texte"] = "PLF"
b.loc[b["PLFSS"] & ~b["PLF"], "texte"] = "PLFSS"
b.loc[b["PLFSS"] & b["PLF"], "texte"] = "les deux"

print("Activités gardées :", len(b))
print(b.groupby(["annee_budget", "texte"]).size().unstack(fill_value=0))

# 3. Qui sont les lobbies ? (type d'organisation)
info = pd.read_csv("Vues_Separees/1_informations_generales.csv", sep=";", low_memory=False)
b = b.merge(info[["representants_id", "denomination", "label_categorie_organisation"]],
            on="representants_id", how="left")
print()
print(b["label_categorie_organisation"].value_counts().head(10))

# 4. Quels secteurs ? (un lobby peut avoir plusieurs secteurs)
sect = pd.read_csv("Vues_Separees/9_secteurs_activites.csv", sep=";")
bs = b.merge(sect, on="representants_id", how="left")
print()
print(bs["secteur_activite"].value_counts().head(15))

# 5. On sauvegarde le tableau propre dans ton dossier
b.to_csv("activites_budget.csv", index=False)
```
